```python
df= pd.read_csv('/retails_sales.csv')
```

```python
import sqlite3

# Create a SQLite database
conn = sqlite3.connect("sales_data.db")

# Save DataFrame to a table named "sales"
df.to_sql("sales", conn, if_exists="replace", index=False)

# Test: Show first few rows using SQL
pd.read_sql_query("SELECT * FROM sales LIMIT 5", conn)
```

```python
# Rename 'quantiy' to 'quantity' for SQL compatibility
df.rename(columns={'quantiy': 'quantity'}, inplace=True)
```

```python
# Create an in-memory SQLite database and write the DataFrame to a table
conn = sqlite3.connect(":memory:")
df.to_sql("sales", conn, index=False, if_exists="replace")
```

```python
# Run the updated SQL query
query = """
SELECT
    category AS product,
    SUM(quantity) AS total_qty,
    SUM(quantity * price_per_unit) AS revenue
FROM sales
GROUP BY category
"""
summary_df = pd.read_sql_query(query, conn)
summary_df
```

```python
# Plot the results
summary_df.plot(kind='bar', x='product', y='revenue', legend=False)
plt.title("Revenue by Product Category")
plt.xlabel("Product Category")
plt.ylabel("Revenue")
plt.tight_layout()
plt.savefig("sales_chart.png")
plt.show()
```
